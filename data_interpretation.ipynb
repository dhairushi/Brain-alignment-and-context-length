```python
import pandas as pd
from pathlib import Path

DATA = Path('/content/drive/MyDrive/Brain_alignment/lpp_data')
word_info = pd.read_csv(DATA / 'annotation/EN/lppEN_word_information.csv')

# Run-15 BOLD: 286 TRs × 2s = 572s
# Check which section has duration closest to 572s
for sec in sorted(word_info['section'].unique()):
    subset = word_info[word_info['section'] == sec]
    duration = subset['offset'].max()
    print(f"Section {sec}: max offset = {duration:.1f}s  (diff from 572s: {abs(duration - 572):.1f}s)")
```

```python
TR = 2.0
SECTION = 8

sec8 = word_info[word_info['section'] == SECTION].copy()
sec8['TR_index'] = (sec8['onset'] / TR).astype(int)
sec8['TR_index'] = sec8['TR_index'].clip(upper=285)

print(f"Words in section 8: {len(sec8)}")
print(f"TR range: {sec8['TR_index'].min()}–{sec8['TR_index'].max()}")
print(sec8[['word', 'onset', 'pos', 'TR_index']].head(10))

sec8.to_csv(DATA / 'word_TR_alignment_run15.csv', index=False)
print("\nSaved: word_TR_alignment_run15.csv")
```

```python
import nibabel as nib
import json

word_info = pd.read_csv(DATA / 'annotation/EN/lppEN_word_information.csv')

for run in ['16', '17', '18']:
    bold_path = FUNC / f'sub-EN057_task-lppEN_run-{run}_echo-1_bold.nii.gz'
    img = nib.load(str(bold_path))
    n_trs = img.shape[3]
    duration = n_trs * 2.0
    print(f"\nRun {run}: {n_trs} TRs = {duration:.1f}s")
    for sec in sorted(word_info['section'].unique()):
        subset = word_info[word_info['section'] == sec]
        diff = abs(subset['offset'].max() - duration)
        print(f"  Section {sec}: diff={diff:.1f}s")
```
